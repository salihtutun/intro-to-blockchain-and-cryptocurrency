## Why we need a shared rule

**Paper IOUs.** Ava's napkin, Ben's sticky note. Cara does not know which
counts. **Uncoordinated private money.** Those markets freeze or crash.

**Monopoly.** If every player prints $500, the game dies. If only the banker
prints, and the box lid names the amount, the game lives.

**Google Form max=30.** Posted rule, shared counter. That *is* a tiny
smart contract.

| Ethereum word | Cafeteria word |
|---|---|
| gas | You paid for the *attempt*, even if the snack did not drop |
| revert | No snack, machine still hummed |
| ERC-20 | Every soda machine uses the **same buttons** so one app works |
| minter | The only banker |
| pauser | Freeze button (compliance / something's wrong) |

**Never mainnet.** Mainnet is a real machine with real cash.

**Slow walk — Monopoly vs napkins.**
If Ava’s napkin says “I owe 500” and Ben’s sticky note says the same, Cara
does not know which counts. Markets freeze. A **shared rule** (the box lid,
or a contract) is how strangers agree on the same numbers.

Gas vs revert, one picture: you put $2 in a vending machine. The coil jams.
The machine hummed (gas spent). The dollar usually comes back in a fair
machine (revert = no snack *and* no state change). On Ethereum the *hum*
still costs you. That is why empty-faucet and wrong-network attempts hurt
even when nothing minted.
