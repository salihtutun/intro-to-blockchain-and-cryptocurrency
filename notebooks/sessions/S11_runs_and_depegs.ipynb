A **peg** says "this chip should stay worth $1."
A **depeg** is the moment the market laughs.
A **run** is everyone hitting redeem at once — a fire drill with one door.



## Words we use today

Read this once. If a later cell uses the word, come back here.

| Word | In this class it means |
|---|---|
| **Peg** | A promise that the chip stays worth a number (often $1, here maybe 0.50 CLASSUSD). |
| **Depeg** | The market price walks away from that number. |
| **Run** | Everyone redeems at once. One door, Saturday rush. |
| **Depth** | How much CLASSUSD is actually in the bowl. Thin bowl → wild quotes. |
| **Contagion** | Your pause or depeg bricks *other* coins that priced through you. |
| **Kill criteria** | If-then with numbers. Not “we’ll watch the market.” |



## Pictures

**Ice-cream gift cards.** 10,000 cards, 200 scoops. Saturday rush →

**School bus stamp.** Hallway price 40 cents = a depeg.

**Class AMM.** CLASSUSD mint pauses, someone empties a bowl. Quotes go
wild. LPs hold a lopsided jar. Payers cannot trust the rail.

**Who gets hurt**
- LPs: the jar is the wrong mix.
- Holders: exit costs a fortune (slippage).
- Payers: the rail is not a known CLASSUSD value.

**Key player vs contagion.** If everyone prices through *your* bowl, you
are a key player. If your pause bricks three coins, you are a contagion
vector. Both can be true.

**Kill criteria** are if-then, not vibes.
Bad: "we'll watch the market."
Good: "If |price − 0.50|/0.50 > 40% and depth < 50 CLASSUSD, pause."

Naranjo *Stablecoins and Peg Risk* is L3.

**Slow walk — gift cards vs scoops.**
10,000 cards, 200 scoops. If everyone ran *tonight*, each card is really
200/10,000 = **0.02 scoops**. The poster still says “1 card = 1 scoop.”
That poster is the peg. The 0.02 is the *run price*.

Saturday: 150 people show up. You honor min(150, 200) = 150. Freezer has 50.
Next wave of 80: you honor only 50. The rest hold a PDF.

A kill rule must be copy-pasteable:
`IF abs(price-peg)/peg > 0.40 AND depth < 50 CLASSUSD THEN pause.`
“We’ll be careful” is not a rule.
