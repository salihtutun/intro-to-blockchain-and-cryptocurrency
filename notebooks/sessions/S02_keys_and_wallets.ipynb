## 1. A hash is a tamper-evident sticker

Gym bag + sticker with a 64-character code. If anyone moves a sock, the
code no longer matches. That sticker is a **hash**.

**Examples**
- Download sites show an SHA-256. Different hash → file was changed.
- Phone update "signed" checks are the same family.
- A blockchain seals each block with a hash of the stuff inside.

Rules of a good hash:
1. Same input → same output, always.
2. Change one letter → unrecognizable new fingerprint.
3. You cannot walk backwards from the fingerprint to the essay.

**Slow walk.** Essay A hashes to `9f3a…`. Essay B is the same essay plus a period.
The new hash looks like `c1e0…` — not “almost 9f3a.” There is no “close.”
That is why a blockchain can notice a quiet edit: the sticker no longer matches.



### Keep this straight

**If you remember one thing.** A hash detects tampering. It does not encrypt the message. Anyone can still *read* the payment.

**Common mix-up.** “Crypto means the payment is secret.” In this class the payment is usually public. The *approval* is locked.

**Why Eagecoin cares.** When we later post an EAGE transfer, the hash of the transaction is the receipt we grade — not a screenshot of a balance.



```python
import hashlib

def sha256(text: str) -> str:
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

# YOUR TURN 1 -- type two messages. Change ONE character in the second.

original = "???"
tweaked = "???"

print("original:", sha256(original))
print("tweaked: ", sha256(tweaked))
print("same?", sha256(original) == sha256(tweaked))

```

## 2. A signature is a yes, not a hidden balance

People hear "crypto" and think spy messages.

In this class the important move is a **signature**:
"The person who knows this secret approved this payment."

**House key.** The lock does not hide how much pizza is in the fridge.
It decides who may open the door.

**Group chat.** Anyone can *read* "Ava sends 2." Only Ava's key should
*authorize* it.

**Bank app.** The bank checks your password, then moves numbers in *their*
list. In a wallet, *you* hold the key the network checks.

**Slow walk — two keys, one locker.**
- **Address (public):** the locker number on the gym wall. Friends can drop CLASSUSD in.
- **Private key / seed:** the only key that opens it. The network never stores this for you.

A **signature** is you turning the key in front of witnesses. They see *that you
approved* “send 2 CLASSUSD to Ben.” They do not learn a hidden pile of cash
from the signature itself. The pile lives on the ledger.



### Keep this straight

**If you remember one thing.** Address = locker number. Seed = house key. Signature = “I approved this.”

**Common mix-up.** Pasting a seed “just for the homework PDF.” That is a security fail, not extra credit.

**Why Eagecoin cares.** The roster needs your *address* so we can airdrop EAGE and CLASSUSD. It must never need your seed.



```python
import hashlib
import hmac

def toy_sign(secret: str, message: str) -> str:
    """TOY lock. Real wallets do not work this way."""
    return hmac.new(secret.encode(), message.encode(), hashlib.sha256).hexdigest()

def toy_verify(secret: str, message: str, signature: str) -> bool:
    return hmac.compare_digest(toy_sign(secret, message), signature)

# YOUR TURN 2 -- try Ava's secret, then Ben's, then a changed amount.

ava_secret = "ava-never-put-this-in-a-pdf"
ben_secret = "ben-also-fake"
payment = "send 2 CLASSUSD from Ava to Ben"
sig = toy_sign(ava_secret, payment)

print("Ava's key works?", ???)          # toy_verify(...)
print("Ben forged it?", ???)
print("Changed amount still valid?", ???)

```
