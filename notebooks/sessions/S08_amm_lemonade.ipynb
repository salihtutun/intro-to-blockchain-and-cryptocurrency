An **Automated Market Maker** is a fishbowl with two kinds of cash.
The rule: **the product of the two piles stays the same.**

\(x \times y = k\)

Nobody has to sit there yelling buy/sell. The bowl *is* the market.



## Words we use today

Read this once. If a later cell uses the word, come back here.

| Word | In this class it means |
|---|---|
| **AMM** | A fishbowl with two piles. The bowl *is* the market. No human market-maker. |
| **Constant product** | x × y = k. After a trade the product stays the same. |
| **Spot price** | y / x — CLASSUSD per one tiny extra TEAM, *right now*. |
| **Average price** | What *you* actually paid for this chunk (paid ÷ team_out). |
| **Slippage** | You moved the bowl, so you paid worse than the starting spot. |
| **LP** | Liquidity provider: the person who *filled* the jar. |
| **Impermanent loss** | Jar vs just holding the two stacks. A snapshot gap, not a fee. |



## Pictures

**Lemonade + dollars.** 10 cups and $10. \(k = 100\).
Take 2 cups out → 8 cups left → dollars become \(100/8 = 12.5\).
You paid **$2.50** for 2 cups. The *next* cup is pricier. That extra
pain is **slippage** — you moved the bowl.

**Trading cards.** Jar has 20 Charizards and 100 commons. Pull too many
Charizards and each one costs a pile of commons.

**School store / LP.** If you *filled* the jar you are the **liquidity
provider**. When prices move, your jar looks different than if you had
**held** the two stacks in your backpack. That gap is **impermanent loss**.

**Spot price** = y / x = CLASSUSD per TEAM (how many dollars for a tiny

**Slow walk — Quiz 2 numbers (write these on paper).**
Start: 100 TEAM, 100 CLASSUSD. \(k = 10{,}000\). Spot = 100/100 = **1**.

Buy 10 TEAM:
- new x = 90
- new y = 10000 / 90 = **111.1111**
- you paid = 111.1111 − 100 = **11.1111** CLASSUSD
- average = 11.1111 / 10 = **1.1111**
- new spot = 111.1111 / 90 = **1.2346**

You paid 1.11 on average. The *next* tiny buy is 1.23. That gap is slippage.
The bowl got TEAM-poor and CLASSUSD-rich, so TEAM is now pricier.

**L2 numbers.** 200 TEAM, 100 CLASSUSD, buy 40:
new x=160, new y=125, paid=25, avg=0.625, spot=0.78125.
Hold-the-stacks at the new spot is worth 256.25; the jar is 250; IL snapshot = **6.25**.



**Picture.** Same Quiz 2 numbers. You slide *along* the curve; the product stays 10,000.



```python
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.4, 3.9),
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# Picture: xy = k. Buying TEAM slides you up the curve — you move the price.
import numpy as np

k = 10_000
x0, y0 = 100.0, 100.0
x1, y1 = 90.0, k / 90.0
xs = np.linspace(40, 180, 240)
ys = k / xs

fig, ax = plt.subplots()
ax.plot(xs, ys, color="#231f1b", lw=2, label="x × y = 10,000")
ax.scatter([x0, x1], [y0, y1], s=70, zorder=3, color="#c45c26")
ax.annotate("start (100, 100)\nspot = 1.00", (x0, y0), xytext=(118, 70),
            arrowprops=dict(arrowstyle="->", color="#666"), fontsize=10)
ax.annotate("after buy 10 TEAM\n(90, 111.11)\nspot = 1.23", (x1, y1), xytext=(45, 145),
            arrowprops=dict(arrowstyle="->", color="#666"), fontsize=10)
ax.set_xlabel("TEAM in the bowl (x)")
ax.set_ylabel("CLASSUSD in the bowl (y)")
ax.set_title("AMM: you paid 11.11; the next sip costs 1.23 — that gap is slippage")
ax.set_xlim(40, 180)
ax.set_ylim(40, 180)
ax.legend(frameon=False, loc="upper right")
plt.tight_layout()
plt.show()

```
