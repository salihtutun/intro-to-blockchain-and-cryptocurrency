## 1. Money has three jobs — not one

Think about a **cafeteria punch card**.

| Job | Fancy name | Cafeteria | Phone |
|---|---|---|---|
| Pay for stuff | medium of exchange | Punch the card, get pizza | Venmo for an Uber |
| Keep value for later | store of value | Save punches for Friday | Money sitting in a bank app |
| Write prices | unit of account | "A slice = 2 punches" | "Hoodie = $45" |

**Example A — lunch club.** 20 students. Each gets 10 punches. Pizza = 2 punches.
If everyone hoards punches "for later," the line dies. People still want pizza.
The *medium of exchange* hid. That is a money problem, not a pizza problem.

**Example B — concert tickets.** Great at job 1 (get in the door). Terrible
everyday money. You cannot buy gum with a stub.

**Example C — group-chat IOUs.** "I owe you a soda" works *inside the chat*.
The gas station does not care.

**Example D — gift card.** Good at the one store. Bad as a country's money.

> Class rule: **do not invent a token until you can name the job it does.**

**Slow walk.** Same cafeteria, three failures:
- Only 8 punches for 30 hungry people → job 1 dies (nothing to *pay* with).
- Punches expire at 2pm → job 2 dies (you cannot save for Friday).
- Every table invents its own “slice price” → job 3 dies (you cannot compare).

A concert ticket fails jobs 2 and 3 on purpose. That is fine for a door.
It is a bad design for *money*.



### Keep this straight

**If you remember one thing.** Name the job first. A token with no job is a sticker.

**Common mix-up.** Calling Bitcoin “money” because it is digital. Digital is a *rail*. Money is the three jobs.

**Why Eagecoin cares.** Eagecoin must say which job it does in *this* classroom (usually: pay for help, or a cookie, or a club tab).

## 3. The babysitting co-op (why this course exists)

## 4. Asset for you, promise for the group

When **you** hold a punch, it feels like treasure (an **asset**).

When the **cafeteria** took that punch, it promised pizza (a **liability**).

Both are true. That is a T-account — the oldest trick in finance.

**Examples**
1. **Gift card** = your asset + the store's IOU.
2. **Bank-app dollar** = your asset + the bank's liability.
3. **Class coin** = your asset + *your team's promise* (minutes, a cookie, an invoice).

If the store dies, the gift card is a pretty PDF. The token did not become pizza.

A MIT idea, said simply: a blockchain and a bank book are both **databases
of transfers**. Money is whatever shows up in that database *a lot*.
